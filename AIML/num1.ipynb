```python
'''Create a DataFrame from a dictionary containing Name and Age.'''

import pandas as pd
data = {"employee" :["sudhi","rajeev","athul"],
        'Age': [24,25,23]
        }

student = pd.DataFrame(data)
print(student)
```

```python
'''Create a DataFrame from a list of lists containing student details'''

list_of_lists = [['Amar', 'Amar nivas mumbai',27], ['Akbar', 'Haji street mumbai',24], ['Anthony', 'Jesus christ road mumbai',25]]
df = pd.DataFrame(list_of_lists, columns = ['Name','Address', 'Age'])
df
```

```python
'''Create a DataFrame from a list of tuples'''

names = ['Arun ','Bharath', 'suraj']
profession = ['IT Employee','Business','Actor']

list_tuple = list(zip(names,profession))
df = pd.DataFrame(list_tuple,columns = ['Name','Profession'])
print(list_tuple)
print(df)
```

```python
'''Create a DataFrame from a NumPy array.'''

import numpy as np
a =np.random.randint(1,10,(4,5))
sf = pd.DataFrame(a, columns =['A','B','C','D','E'])
sf
```

```python
'''Create a DataFrame with custom row labels.'''
data = {
    "Name": ["John", "Emma", "Alice"],
    "Age": [22, 21, 23]
}

df = pd.DataFrame(data, index=["A", "B", "C"])
print(df)
```

```python
'''Create an empty DataFrame.'''

df = pd.DataFrame()
print(df)
```

```python
'''Create a DataFrame with only one column named Marks.'''
df = pd.DataFrame({
    "Marks": [85, 90, 78, 88]
})

print(df)
```

```python
'''Create a DataFrame using two Pandas Series.'''

name = pd.Series(["John", "Emma", "Alice"])
age = pd.Series([22, 21, 23])

df = pd.DataFrame({
    "Name": name,
    "Age": age
})

print(df)
```

```python
'''Create a DataFrame with three columns: ID, Name, and Department'''

data = {
    "ID": [101, 102, 103],
    "Name": ["John", "Emma", "Alice"],
    "Department": ["CSE", "ECE", "ME"]
}

df = pd.DataFrame(data)
print(df)
```

```python
'''Create a DataFrame containing five students and their marks'''

data = {
    "Student": ["John", "Emma", "Alice", "David", "Bob"],
    "Marks": [85, 90, 78, 88, 92]
}

df = pd.DataFrame(data)
print(df)
```

```python
'''Create a DataFrame with 10 rows and 4 columns filled with zeros.'''
df = pd.DataFrame(
    np.zeros((10, 4)),
    columns=["A", "B", "C", "D"]
)

print(df)
```

```python
'''Create a DataFrame with 5 rows and 3 columns filled with ones.'''

df = pd.DataFrame(
    np.ones((5, 3)),
    columns=["A", "B", "C"]
)

print(df)
```

```python
'''Create a DataFrame with values from 1 to 20 arranged into 5 rows and 4 columns.'''

data = np.arange(1, 21).reshape(5, 4)

df = pd.DataFrame(
    data,
    columns=["A", "B", "C", "D"]
)

print(df)
```

```python
'''Create a DataFrame with mixed data types (integer, string, float, boolean).'''

data = {
    "ID": [101, 102, 103],
    "Name": ["John", "Emma", "Alice"],
    "Percentage": [85.5, 90.0, 78.25],
    "Passed": [True, True, False]
}

df = pd.DataFrame(data)

print(df)
```
