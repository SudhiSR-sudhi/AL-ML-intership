```python
'''Add a new column DiscountedSales = TotalSales × 0.9 (10% discount).'''

df["DiscountedSales"] = df["TotalSales"] *0.9
print(df)
```

```python
'''Find the total sales per Region.'''

total = df.groupby("Region")["TotalSales"].sum()
print(total)
```

```python
'''Find the average Units sold per Product.'''
avarege = df.groupby("Product")["Units"].mean()
print(avarege)
```

```python
'''Count how many sales each SalesPerson made.'''

sales_person = df.groupby("SalesPerson")["Units"].sum()
print(sales_person)
```

```python
'''Find the highest TotalSales for each Region.'''

print("Highest TotalSales:", df["TotalSales"].max())
person = df.loc[df["TotalSales"].idxmax(), "SalesPerson"]
print(person)
```

```python
'''Find the total revenue per Product'''

revenue_summary = df.groupby('Product')[['TotalSales', 'DiscountedSales']].sum()
print(revenue_summary)
```

```python
'''Identify the top-performing SalesPerson in terms of revenue.'''

sales_person = df.groupby("SalesPerson")["Units"].sum()
top_name = df.loc[df["SalesPerson"].idxmax(), "Student Name"]
print(top_name)
```

```python
'''Find the product that generated the highest overall sales'''

product_sales = df.groupby("Product")["TotalSales"].sum().reset_index()
top_product = product_sales.sort_values(by="TotalSales", ascending=False)
print(top_product)
```
