```python

job_df.groupby("employment_type")['salary_usd'].value_counts()

#Here we conclude that we grouped the employee type and the salary in usd and the value counts of the salaey in usd 
```

```python
#Average salary by experience level


c = job_df.groupby("experience_level")["salary_usd"].mean()
plt.figure(figsize=(10,5))
plt.barh(c.index, c.values, color="brown")
plt.title("Average salary by experience level")
plt.xlabel("salary")
plt.ylabel("experience")
plt.show()

#first we have grouped the experience level and salary in usd and calculate the mean salary
#then we have plot bar horizontal graph to show the highest paying salary accourind to the expeience level 
#Here we conclude that accourding to the datasets executive has more salary to the expeience level and low salary should be to the entry level 
```

```python
a = job_df["employee_residence"].value_counts()
plt.figure(figsize=(7, 7))
plt.pie(a, labels=a.index, autopct='%1.1f%%', startangle=140)
plt.title('Employee Residence')
plt.legend()
plt.show()

'''Here we can say the we have count the employee residence of all the employees and plot the graph using pie chart 
and add title as Employee Residence. The below pie chart represents the country where employeees are residenece 
Here the conclution that  most of the employees are from USA = 35.9% and least from Singapore(SG) =2.4% '''
```

```python
job_df.groupby("employment_type")['education_level'].value_counts()


plt.figure(figsize=(10,6))

sns.countplot(data=job_df,x="employment_type",hue="education_level")

plt.title("Education Level by Employment Type")
plt.xlabel("Employment Type")
plt.ylabel("Count")
plt.legend(title="Education Level")
plt.show()


'''Here we group the employee type and education level and 
conclude that full time education level has more employement type and bachelors has more job that others'''
```

```python
cross = pd.crosstab(job_df["employment_type"],job_df["education_level"])
sns.heatmap(cross,annot=True,cmap="Blues",fmt="d")
plt.title("Employment Type vs Education Level")
plt.show()

'''Here we have same data as above using heatmap'''
```
